### `append()`: Adds an item to the end of the list.

```python
my_list = [1, 2, 3]
my_list.append(4)
print(f"After append(4): {my_list}")
```

### `extend()`: Extends the list by appending all the items from an iterable.

```python
my_list = [1, 2, 3]
my_list.extend([4, 5])
print(f"After extend([4, 5]): {my_list}")
```

### `insert(index, element)`: Inserts an item at a specified position.

```python
my_list = [1, 2, 3]
my_list.insert(1, 'new_item')
print(f"After insert(1, 'new_item'): {my_list}")
```

### `remove(element)`: Removes the first occurrence of a specified item.

```python
my_list = [1, 2, 'new_item', 3]
my_list.remove('new_item')
print(f"After remove('new_item'): {my_list}")
```

### `pop(index)`: Removes and returns the item at a given index. If no index is specified, `pop()` removes and returns the last item.

```python
my_list = [1, 2, 3, 4]
removed_item_index = my_list.pop(1)
print(f"After pop(1): {my_list}, removed item: {removed_item_index}")
removed_item_last = my_list.pop()
print(f"After pop(): {my_list}, removed item: {removed_item_last}")
```

### `clear()`: Removes all items from the list.

```python
my_list = [1, 2, 3]
my_list.clear()
print(f"After clear(): {my_list}")
```

### `index(element, start, end)`: Returns the index of the first occurrence of a specified item. Optional `start` and `end` parameters can define a search range.

```python
my_list = [10, 20, 30, 20, 40]
index_of_20 = my_list.index(20)
print(f"Index of 20: {index_of_20}")
index_of_20_after_first = my_list.index(20, 2) # Search from index 2
print(f"Index of 20 (starting from index 2): {index_of_20_after_first}")
```

### `count(element)`: Returns the number of times a specified item appears in the list.

```python
my_list = [1, 2, 2, 3, 2, 4]
count_of_2 = my_list.count(2)
print(f"Count of 2: {count_of_2}")
```

### `sort(key=None, reverse=False)`: Sorts the items of the list in ascending order by default. `reverse=True` for descending order.

```python
my_list = [3, 1, 4, 1, 5, 9, 2]
my_list.sort()
print(f"After sort(): {my_list}")
my_list.sort(reverse=True)
print(f"After sort(reverse=True): {my_list}")

words = ['banana', 'apple', 'cherry']
words.sort(key=len) # Sort by length of string
print(f"After sort by length: {words}")
```

### `reverse()`: Reverses the order of items in the list in-place.

```python
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print(f"After reverse(): {my_list}")
```

### `copy()`: Returns a shallow copy of the list.

```python
original_list = [1, 2, 3]
copied_list = original_list.copy()
print(f"Original list: {original_list}, Copied list: {copied_list}")
copied_list.append(4)
print(f"Original list after appending to copy: {original_list}")
print(f"Copied list after appending: {copied_list}")
```

### `len(list)`: Built-in function to get the number of items in the list.

```python
my_list = [1, 2, 3, 4, 5]
length = len(my_list)
print(f"Length of the list: {length}")
```

### `max(list)`: Built-in function to get the largest item in the list.

```python
my_list = [10, 5, 20, 15]
maximum = max(my_list)
print(f"Maximum item in the list: {maximum}")
```

### `min(list)`: Built-in function to get the smallest item in the list.

```python
my_list = [10, 5, 20, 15]
minimum = min(my_list)
print(f"Minimum item in the list: {minimum}")
```

### `sum(list)`: Built-in function to get the sum of all numeric items in the list.

```python
my_list = [10, 5, 20, 15]
total_sum = sum(my_list)
print(f"Sum of items in the list: {total_sum}")
```
